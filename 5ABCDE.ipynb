```python
import matplotlib.pyplot as plt
x=[11,22,33,44]
y=[40,66,95,88]
plt.plot(x,y)
plt.title("line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.savefig("line_plot.png")
plt.show()
```

```python
import pandas as pd
df=pd.DataFrame({'category':['1','2','3'],'values':[10.7,20.5,15.1]})
df.plot(kind='bar',x='category',y='values')
plt.title("BarPlot")
plt.show()
```

```python
import seaborn as sns
import numpy as np
data=np.random.randn(1000)
sns.histplot(data,kde=True)
plt.title("Histogram & Density")
plt.show()
```

```python
x=np.random.rand(50)
y=np.random.rand(50)
plt.scatter(x,y)
plt.title("Scatter plot")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
sns.boxplot(data=np.random.randn(10014))
plt.title("Box plot")
plt.show()
```
